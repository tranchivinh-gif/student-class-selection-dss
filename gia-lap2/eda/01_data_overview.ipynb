```python
# Đọc dữ liệu Excel
file_path = "../data/data-clean/du_lieu_that.csv"
raw_df = lib.pd.read_csv(file_path)

choice_columns = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]
dfs = {
    f"Diem_L{grade}": raw_df[
        ["ma_hoc_sinh"]
        + [col for col in raw_df.columns if col.startswith(f"L{grade}_")]
        + choice_columns
    ].copy()
    for grade in [6, 7, 8, 9]
}

# Truy cập dữ liệu
df_l6 = dfs["Diem_L6"]
df_l7 = dfs["Diem_L7"]
df_l8 = dfs["Diem_L8"]
df_l9 = dfs["Diem_L9"]

# Hiển thị
display(df_l6.head())
display(df_l7.head())
display(df_l8.head())
display(df_l9.head())
```

```python
# Xác định kích thước dữ liệu của từng sheet
for sheet_name, df in dfs.items():
    print(f"Sheet: {sheet_name}")
    print(f"  Số dòng: {df.shape[0]:,}")
    print(f"  Số cột: {df.shape[1]}")
    print(f"  Kích thước: {df.shape}")
    print("-" * 40)
```

```python
# Kiểm tra tên và kiểu dữ liệu từng cột
for sheet_name, df in dfs.items():
    print(f"\n{'=' * 60}")
    print(f"SHEET: {sheet_name}")
    print(f"{'=' * 60}")

    column_info = lib.pd.DataFrame({
        "Tên cột": df.columns,
        "Kiểu dữ liệu": df.dtypes.values
    })

    display(column_info)
```

```python
# Kiểm tra tính liên kết mã học sinh giữa các lớp 6, 7, 8, 9

id_col = "ma_hoc_sinh"

# 1. Kiểm tra mã học sinh bị thiếu hoặc trùng trong từng sheet
for sheet_name, df in dfs.items():
    print(f"\nSHEET: {sheet_name}")
    print(f"  Tổng số học sinh: {len(df):,}")
    print(f"  Mã học sinh thiếu: {df[id_col].isna().sum():,}")
    print(f"  Mã học sinh trùng: {df[id_col].duplicated().sum():,}")
    print(f"  Mã học sinh duy nhất: {df[id_col].nunique():,}")

# 2. So sánh mã học sinh giữa các lớp
id_sets = {
    sheet_name: set(df[id_col].dropna())
    for sheet_name, df in dfs.items()
}

common_ids = set.intersection(*id_sets.values())
all_ids = set.union(*id_sets.values())

print("\n" + "=" * 60)
print("KIỂM TRA LIÊN KẾT GIỮA CÁC LỚP")
print("=" * 60)

print(f"Tổng số mã học sinh khác nhau: {len(all_ids):,}")
print(f"Số học sinh xuất hiện ở cả 4 lớp: {len(common_ids):,}")

for sheet_name, ids in id_sets.items():
    missing_ids = all_ids - ids
    print(f"{sheet_name}: Thiếu {len(missing_ids):,} mã học sinh")

# 3. Kiểm tra tính nhất quán toàn bộ
is_consistent = (
    all(ids == common_ids for ids in id_sets.values())
    and all(
        df[id_col].notna().all()
        and df[id_col].is_unique
        for df in dfs.values()
    )
)

print("\nKẾT LUẬN:")
if is_consistent:
    print("PASS: Mã học sinh liên kết nhất quán giữa cả 4 lớp.")
else:
    print("FAIL: Phát hiện mã học sinh thiếu, trùng hoặc không khớp giữa các lớp.")
```

```python
# Kiểm tra lựa chọn 9 môn của từng học sinh có nhất quán qua lớp 6–9

id_col = "ma_hoc_sinh"

choice_cols = [
    "Vật lí",
    "Hóa học",
    "Sinh học",
    "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học",
    "Công nghệ",
    "Âm nhạc",
    "Mĩ thuật"
]

# Ghép dữ liệu lựa chọn môn học theo mã học sinh
merged = dfs["Diem_L6"][[id_col] + choice_cols].copy()

for grade in [7, 8, 9]:
    merged = merged.merge(
        dfs[f"Diem_L{grade}"][[id_col] + choice_cols],
        on=id_col,
        how="inner",
        suffixes=("", f"_L{grade}"),
        validate="one_to_one"
    )

# So sánh từng môn giữa lớp 6 và các lớp còn lại
comparison = lib.pd.DataFrame(index=merged.index)

for grade in [7, 8, 9]:
    comparison[f"L6_vs_L{grade}"] = (
        merged[choice_cols].to_numpy()
        == merged[[f"{col}_L{grade}" for col in choice_cols]].to_numpy()
    ).all(axis=1)

# Học sinh có lựa chọn nhất quán qua 4 lớp
consistent = comparison.all(axis=1)

print("=" * 60)
print("KIỂM TRA TÍNH NHẤT QUÁN LỰA CHỌN MÔN HỌC")
print("=" * 60)

print(f"Số học sinh được đối chiếu: {len(merged):,}")
print(f"Số học sinh nhất quán: {consistent.sum():,}")
print(f"Số học sinh không nhất quán: {(~consistent).sum():,}")
print(f"Tỷ lệ nhất quán: {consistent.mean() * 100:.2f}%")

for col in comparison.columns:
    print(f"{col}: {(~comparison[col]).sum():,} học sinh khác lựa chọn")

# Hiển thị học sinh có lựa chọn không nhất quán
if (~consistent).any():
    print("\nDanh sách học sinh có lựa chọn không nhất quán:")
    display(
        lib.pd.concat(
            [
                merged.loc[~consistent, [id_col]],
                comparison.loc[~consistent]
            ],
            axis=1
        )
    )
else:
    print("\nPASS: Tất cả học sinh có lựa chọn môn học nhất quán qua 4 lớp.")
```
