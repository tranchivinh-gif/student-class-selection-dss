```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import src.imports as lib
```

```python
# Đọc dữ liệu Excel
file_path = "../data/data-clean/du_lieu_that.csv"
raw_df = lib.pd.read_csv(file_path)

choice_columns = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]
dfs = {
    f"Diem_L{grade}": raw_df[
        ["ma_hoc_sinh"]
        + [col for col in raw_df.columns if col.startswith(f"L{grade}_")]
        + choice_columns
    ].copy()
    for grade in [6, 7, 8, 9]
}

# Truy cập dữ liệu
df_l6 = dfs["Diem_L6"]
df_l7 = dfs["Diem_L7"]
df_l8 = dfs["Diem_L8"]
df_l9 = dfs["Diem_L9"]

# Hiển thị
display(df_l6.head())
display(df_l7.head())
display(df_l8.head())
display(df_l9.head())
```

```python
# ==========================================
# LEARNING TRAJECTORY ANALYSIS
# XU HƯỚNG ĐIỂM TRUNG BÌNH LỚP 6 → LỚP 9
# ==========================================

fig_dir = lib.Path("figures/learning_trajectory")
report_dir = lib.Path("reports/learning_trajectory")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

results = []

for grade in [6, 7, 8, 9]:

    sheet_name = f"Diem_L{grade}"
    df = dfs[sheet_name]

    # Lấy 8 cột điểm cả năm
    score_cols = [
        col for col in df.columns
        if col.endswith("_ca_nam")
    ]

    # Trung bình 8 môn của từng học sinh
    student_mean = df[score_cols].mean(axis=1)

    # Thống kê theo khối lớp
    results.append({
        "Lớp": grade,
        "Số học sinh": student_mean.count(),
        "Mean": student_mean.mean(),
        "Median": student_mean.median(),
        "Std": student_mean.std(),
        "Min": student_mean.min(),
        "Max": student_mean.max()
    })

# ==========================================
# 1. BẢNG THỐNG KÊ
# ==========================================

trend_df = lib.pd.DataFrame(results).round(4)

display(trend_df)

trend_df.to_csv(
    report_dir / "average_score_trend.csv",
    index=False,
    encoding="utf-8-sig"
)

# ==========================================
# 2. LINE PLOT - XU HƯỚNG ĐIỂM TRUNG BÌNH
# ==========================================

fig, ax = lib.plt.subplots(figsize=(10, 6))

lib.sns.lineplot(
    data=trend_df,
    x="Lớp",
    y="Mean",
    marker="o",
    markersize=9,
    linewidth=2.5,
    color="steelblue",
    ax=ax
)

# Hiển thị giá trị trên từng điểm
for _, row in trend_df.iterrows():
    ax.annotate(
        f'{row["Mean"]:.3f}',
        (row["Lớp"], row["Mean"]),
        textcoords="offset points",
        xytext=(0, 12),
        ha="center",
        fontsize=11
    )

ax.set_title(
    "Xu hướng điểm trung bình từ lớp 6 đến lớp 9",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Khối lớp")
ax.set_ylabel("Điểm trung bình 8 môn")
ax.set_xticks([6, 7, 8, 9])
ax.set_ylim(0, 10)
ax.grid(True, linestyle="--", alpha=0.4)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "average_score_trend.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# LEARNING TRAJECTORY ANALYSIS
# SỰ THAY ĐỔI ĐIỂM TỪNG MÔN QUA LỚP 6 → 9
# ==========================================

fig_dir = lib.Path("figures/learning_trajectory")
report_dir = lib.Path("reports/learning_trajectory")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

subject_names = {
    "van": "Ngữ văn",
    "toan": "Toán",
    "ngoai_ngu_1": "Ngoại ngữ 1",
    "gdcd": "GDCD",
    "lich_su_dia_li": "Lịch sử và Địa lí",
    "khtn": "Khoa học tự nhiên",
    "cong_nghe": "Công nghệ",
    "tin_hoc": "Tin học"
}

periods = {
    "HKI": "_hki",
    "HKII": "_hkii",
    "CaNam": "_ca_nam"
}

results = []

# ==========================================
# 1. TÍNH ĐIỂM TRUNG BÌNH TỪNG MÔN THEO LỚP
# ==========================================

for grade in [6, 7, 8, 9]:

    df = dfs[f"Diem_L{grade}"]

    for subject, subject_name in subject_names.items():

        for period, suffix in periods.items():

            col = f"L{grade}_{subject}{suffix}"

            if col not in df.columns:
                continue

            results.append({
                "Lớp": grade,
                "Môn học": subject_name,
                "Học kỳ": period,
                "Mean": df[col].mean(),
                "Median": df[col].median(),
                "Std": df[col].std()
            })

trend_df = lib.pd.DataFrame(results).round(4)

display(trend_df)

# Lưu bảng thống kê
trend_df.to_csv(
    report_dir / "subject_score_trend.csv",
    index=False,
    encoding="utf-8-sig"
)

# ==========================================
# 2. LINE PLOT - XU HƯỚNG TỪNG MÔN
# THEO HKI, HKII VÀ CẢ NĂM
# ==========================================

for period in periods.keys():

    plot_df = trend_df[
        trend_df["Học kỳ"] == period
    ]

    fig, ax = lib.plt.subplots(figsize=(13, 7))

    lib.sns.lineplot(
        data=plot_df,
        x="Lớp",
        y="Mean",
        hue="Môn học",
        style="Môn học",
        markers=True,
        dashes=False,
        linewidth=2,
        markersize=8,
        ax=ax
    )

    ax.set_title(
        f"Xu hướng điểm trung bình từng môn - {period}",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Khối lớp")
    ax.set_ylabel("Điểm trung bình")
    ax.set_xticks([6, 7, 8, 9])
    ax.set_ylim(0, 10)

    ax.legend(
        title="Môn học",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    ax.grid(True, linestyle="--", alpha=0.4)

    lib.plt.tight_layout()

    lib.plt.savefig(
        fig_dir / f"subject_score_trend_{period}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

# ==========================================
# 3. MỨC THAY ĐỔI ĐIỂM LỚP 6 → 9
# ==========================================

annual_df = trend_df[
    trend_df["Học kỳ"] == "CaNam"
]

pivot_df = annual_df.pivot(
    index="Môn học",
    columns="Lớp",
    values="Mean"
)

pivot_df["Thay đổi 6-7"] = pivot_df[7] - pivot_df[6]
pivot_df["Thay đổi 7-8"] = pivot_df[8] - pivot_df[7]
pivot_df["Thay đổi 8-9"] = pivot_df[9] - pivot_df[8]
pivot_df["Thay đổi 6-9"] = pivot_df[9] - pivot_df[6]

pivot_df = pivot_df.round(4)

print("MỨC THAY ĐỔI ĐIỂM TRUNG BÌNH TỪNG MÔN:")
display(pivot_df)

pivot_df.to_csv(
    report_dir / "subject_score_changes.csv",
    encoding="utf-8-sig"
)

# ==========================================
# 4. HEATMAP - MỨC THAY ĐỔI ĐIỂM
# ==========================================

change_cols = [
    "Thay đổi 6-7",
    "Thay đổi 7-8",
    "Thay đổi 8-9",
    "Thay đổi 6-9"
]

fig, ax = lib.plt.subplots(figsize=(11, 7))

lib.sns.heatmap(
    pivot_df[change_cols],
    annot=True,
    fmt=".3f",
    cmap="RdBu_r",
    center=0,
    linewidths=0.5,
    cbar_kws={"label": "Chênh lệch điểm"},
    ax=ax
)

ax.set_title(
    "Mức thay đổi điểm trung bình từng môn qua các năm",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Giai đoạn")
ax.set_ylabel("Môn học")

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "subject_score_change_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# LEARNING TRAJECTORY ANALYSIS
# MỨC ĐỘ ỔN ĐỊNH KẾT QUẢ HỌC TẬP LỚP 6 → 9
# ==========================================

fig_dir = lib.Path("figures/learning_trajectory")
report_dir = lib.Path("reports/learning_trajectory")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

# ==========================================
# 1. TÍNH ĐIỂM TRUNG BÌNH 8 MÔN CỦA TỪNG HỌC SINH
# ==========================================

student_scores = None

for grade in [6, 7, 8, 9]:

    df = dfs[f"Diem_L{grade}"]

    score_cols = [
        col for col in df.columns
        if col.endswith("_ca_nam")
    ]

    temp = lib.pd.DataFrame({
        "ma_hoc_sinh": df["ma_hoc_sinh"],
        f"Lop_{grade}": df[score_cols].mean(axis=1)
    })

    if student_scores is None:
        student_scores = temp
    else:
        student_scores = student_scores.merge(
            temp,
            on="ma_hoc_sinh",
            how="inner",
            validate="one_to_one"
        )

grade_cols = ["Lop_6", "Lop_7", "Lop_8", "Lop_9"]

# ==========================================
# 2. TÍNH CÁC CHỈ SỐ ỔN ĐỊNH
# ==========================================

student_scores["Mean_4_nam"] = student_scores[grade_cols].mean(axis=1)

student_scores["Std_4_nam"] = student_scores[grade_cols].std(
    axis=1, ddof=0
)

student_scores["Range_4_nam"] = (
    student_scores[grade_cols].max(axis=1)
    - student_scores[grade_cols].min(axis=1)
)

student_scores["Delta_6_9"] = (
    student_scores["Lop_9"] - student_scores["Lop_6"]
)

student_scores["Mean_Absolute_Change"] = (
    student_scores[grade_cols]
    .diff(axis=1)
    .iloc[:, 1:]
    .abs()
    .mean(axis=1)
)

# ==========================================
# 3. THỐNG KÊ MÔ TẢ
# ==========================================

stability_cols = [
    "Mean_4_nam",
    "Std_4_nam",
    "Range_4_nam",
    "Delta_6_9",
    "Mean_Absolute_Change"
]

stability_summary = student_scores[stability_cols].describe().round(4)

print("THỐNG KÊ MỨC ĐỘ ỔN ĐỊNH KẾT QUẢ HỌC TẬP")
display(stability_summary)

# ==========================================
# 4. MA TRẬN TƯƠNG QUAN ĐIỂM GIỮA CÁC NĂM
# ==========================================

corr_matrix = student_scores[grade_cols].corr(method="pearson")

print("\nMA TRẬN TƯƠNG QUAN ĐIỂM GIỮA CÁC LỚP")
display(corr_matrix.round(4))

fig, ax = lib.plt.subplots(figsize=(8, 6))

lib.sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".3f",
    cmap="Blues",
    vmin=0,
    vmax=1,
    square=True,
    linewidths=0.5,
    ax=ax
)

ax.set_title(
    "Tương quan kết quả học tập giữa các lớp 6–9",
    fontsize=14,
    fontweight="bold"
)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "stability_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 5. PHÂN PHỐI ĐỘ LỆCH CHUẨN QUA 4 NĂM
# ==========================================

fig, ax = lib.plt.subplots(figsize=(10, 6))

lib.sns.histplot(
    data=student_scores,
    x="Std_4_nam",
    bins=40,
    kde=True,
    color="steelblue",
    ax=ax
)

ax.axvline(
    student_scores["Std_4_nam"].median(),
    color="red",
    linestyle="--",
    label=f'Median = {student_scores["Std_4_nam"].median():.3f}'
)

ax.set_title(
    "Phân phối độ biến động điểm trung bình qua 4 năm",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Độ lệch chuẩn điểm trung bình 4 năm")
ax.set_ylabel("Số học sinh")
ax.legend()

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "stability_std_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 6. BOXPLOT - BIẾN ĐỘNG ĐIỂM GIỮA CÁC LỚP
# ==========================================

long_df = student_scores.melt(
    id_vars="ma_hoc_sinh",
    value_vars=grade_cols,
    var_name="Lớp",
    value_name="Điểm trung bình"
)

fig, ax = lib.plt.subplots(figsize=(10, 6))

lib.sns.boxplot(
    data=long_df,
    x="Lớp",
    y="Điểm trung bình",
    order=grade_cols,
    color="skyblue",
    ax=ax
)

ax.set_title(
    "Phân phối điểm trung bình học sinh qua lớp 6–9",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Khối lớp")
ax.set_ylabel("Điểm trung bình")
ax.set_ylim(0, 10)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "stability_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 7. LƯU KẾT QUẢ PHÂN TÍCH
# ==========================================

student_scores.to_csv(
    report_dir / "student_learning_stability.csv",
    index=False,
    encoding="utf-8-sig"
)

stability_summary.to_csv(
    report_dir / "learning_stability_summary.csv",
    encoding="utf-8-sig"
)

corr_matrix.to_csv(
    report_dir / "learning_stability_correlation.csv",
    encoding="utf-8-sig"
)

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# LEARNING TRAJECTORY ANALYSIS
# PHÂN NHÓM XU HƯỚNG HỌC TẬP: TĂNG, GIẢM, ỔN ĐỊNH
# ==========================================

fig_dir = lib.Path("figures/learning_trajectory")
report_dir = lib.Path("reports/learning_trajectory")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

# Sử dụng dữ liệu đã tính ở bước trước
df_trend = student_scores.copy()

grade_cols = ["Lop_6", "Lop_7", "Lop_8", "Lop_9"]

# ==========================================
# 1. PHÂN NHÓM THEO MỨC THAY ĐỔI LỚP 6 → 9
# ==========================================

# Ngưỡng thay đổi: 0.5 điểm (giả định phân tích)
threshold = 0.5

df_trend["Delta_6_9"] = (
    df_trend["Lop_9"] - df_trend["Lop_6"]
)

conditions = [
    df_trend["Delta_6_9"] > threshold,
    df_trend["Delta_6_9"] < -threshold
]

choices = ["Tăng", "Giảm"]

df_trend["Nhom_xu_huong"] = lib.np.select(
    conditions,
    choices,
    default="Ổn định"
)

# ==========================================
# 2. THỐNG KÊ SỐ LƯỢNG VÀ TỶ LỆ
# ==========================================

summary = (
    df_trend["Nhom_xu_huong"]
    .value_counts()
    .reindex(["Tăng", "Giảm", "Ổn định"], fill_value=0)
    .rename_axis("Nhóm xu hướng")
    .reset_index(name="Số học sinh")
)

summary["Tỷ lệ (%)"] = (
    summary["Số học sinh"] / len(df_trend) * 100
).round(2)

print("PHÂN BỐ NHÓM XU HƯỚNG HỌC TẬP")
display(summary)

# ==========================================
# 3. BAR CHART - PHÂN BỐ NHÓM XU HƯỚNG
# ==========================================

fig, ax = lib.plt.subplots(figsize=(9, 6))

lib.sns.barplot(
    data=summary,
    x="Nhóm xu hướng",
    y="Số học sinh",
    order=["Tăng", "Giảm", "Ổn định"],
    color="steelblue",
    ax=ax
)

for i, row in summary.iterrows():
    ax.text(
        i,
        row["Số học sinh"] + max(summary["Số học sinh"]) * 0.01,
        f'{row["Số học sinh"]:,}\n({row["Tỷ lệ (%)"]:.2f}%)',
        ha="center",
        fontsize=10
    )

ax.set_title(
    "Phân nhóm xu hướng kết quả học tập lớp 6–9",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Nhóm xu hướng")
ax.set_ylabel("Số học sinh")
ax.set_ylim(0, summary["Số học sinh"].max() * 1.2)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "learning_trend_groups.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 4. LINE PLOT - QUỸ ĐẠO TRUNG BÌNH TỪNG NHÓM
# ==========================================

group_means = (
    df_trend.groupby("Nhom_xu_huong")[grade_cols]
    .mean()
    .reindex(["Tăng", "Giảm", "Ổn định"])
)

plot_df = (
    group_means.reset_index()
    .melt(
        id_vars="Nhom_xu_huong",
        value_vars=grade_cols,
        var_name="Lớp",
        value_name="Điểm trung bình"
    )
)

plot_df["Lớp"] = (
    plot_df["Lớp"].str.replace("Lop_", "", regex=False).astype(int)
)

fig, ax = lib.plt.subplots(figsize=(10, 6))

lib.sns.lineplot(
    data=plot_df,
    x="Lớp",
    y="Điểm trung bình",
    hue="Nhom_xu_huong",
    style="Nhom_xu_huong",
    markers=True,
    dashes=False,
    linewidth=2.5,
    markersize=9,
    ax=ax
)

ax.set_title(
    "Quỹ đạo điểm trung bình theo nhóm xu hướng",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Khối lớp")
ax.set_ylabel("Điểm trung bình 8 môn")
ax.set_xticks([6, 7, 8, 9])
ax.set_ylim(0, 10)
ax.legend(title="Nhóm xu hướng")
ax.grid(True, linestyle="--", alpha=0.4)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "learning_trend_trajectories.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 5. LƯU KẾT QUẢ
# ==========================================

df_trend.to_csv(
    report_dir / "student_learning_trend_groups.csv",
    index=False,
    encoding="utf-8-sig"
)

summary.to_csv(
    report_dir / "learning_trend_group_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

group_means.to_csv(
    report_dir / "learning_trend_group_means.csv",
    encoding="utf-8-sig"
)

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# LEARNING TRAJECTORY ANALYSIS
# SO SÁNH QUỸ ĐẠO HỌC TẬP GIỮA CÁC NHÓM NHÃN 0/1
# ==========================================

fig_dir = lib.Path("figures/learning_trajectory")
report_dir = lib.Path("reports/learning_trajectory")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

grade_cols = ["Lop_6", "Lop_7", "Lop_8", "Lop_9"]

# ==========================================
# 1. GHÉP ĐIỂM TRUNG BÌNH 4 NĂM VỚI 9 NHÃN
# ==========================================

# student_scores đã được tạo ở bước phân tích độ ổn định
df_analysis = student_scores[
    ["ma_hoc_sinh"] + grade_cols
].copy()

# Nhãn giống nhau qua 4 lớp, chỉ cần lấy từ lớp 9
labels = dfs["Diem_L9"][
    ["ma_hoc_sinh"] + choice_cols
].copy()

df_analysis = df_analysis.merge(
    labels,
    on="ma_hoc_sinh",
    how="inner",
    validate="one_to_one"
)

# Chuyển dữ liệu điểm từ dạng wide sang long
df_long = df_analysis.melt(
    id_vars=["ma_hoc_sinh"] + choice_cols,
    value_vars=grade_cols,
    var_name="Lớp",
    value_name="Điểm trung bình"
)

df_long["Lớp"] = (
    df_long["Lớp"]
    .str.replace("Lop_", "", regex=False)
    .astype(int)
)

# ==========================================
# 2. TÍNH QUỸ ĐẠO ĐIỂM TRUNG BÌNH THEO NHÃN
# ==========================================

trajectory_reports = []

fig, axes = lib.plt.subplots(
    3, 3,
    figsize=(18, 15)
)

axes = axes.flatten()

for i, label_col in enumerate(choice_cols):

    ax = axes[i]

    plot_data = df_long[
        df_long[label_col].isin([0, 1])
    ].copy()

    # Thống kê điểm trung bình theo lớp và nhóm nhãn
    summary = (
        plot_data
        .groupby(["Lớp", label_col])["Điểm trung bình"]
        .agg(["count", "mean", "median", "std"])
        .reset_index()
    )

    summary["Môn lựa chọn"] = label_col

    summary = summary.rename(
        columns={
            label_col: "Nhãn",
            "count": "Số học sinh",
            "mean": "Mean",
            "median": "Median",
            "std": "Std"
        }
    )

    trajectory_reports.append(summary)

    # Vẽ quỹ đạo trung bình của nhóm 0 và 1
    lib.sns.lineplot(
        data=plot_data,
        x="Lớp",
        y="Điểm trung bình",
        hue=label_col,
        hue_order=[0, 1],
        estimator="mean",
        errorbar=None,
        marker="o",
        linewidth=2.5,
        markersize=7,
        palette={0: "steelblue", 1: "darkorange"},
        ax=ax
    )

    ax.set_title(
        label_col,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_xlabel("Khối lớp")
    ax.set_ylabel("Điểm trung bình 8 môn")
    ax.set_xticks([6, 7, 8, 9])
    ax.set_ylim(0, 10)
    ax.grid(True, linestyle="--", alpha=0.3)

    handles, _ = ax.get_legend_handles_labels()

    ax.legend(
        handles=handles,
        labels=["Không chọn (0)", "Có chọn (1)"],
        title="Nhóm nhãn",
        fontsize=9
    )

fig.suptitle(
    "So sánh quỹ đạo học tập giữa nhóm chọn và không chọn môn",
    fontsize=16,
    fontweight="bold"
)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "learning_trajectory_by_labels.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 3. SO SÁNH MỨC THAY ĐỔI ĐIỂM LỚP 6 → 9
# ==========================================

df_analysis["Delta_6_9"] = (
    df_analysis["Lop_9"] - df_analysis["Lop_6"]
)

change_reports = []

for label_col in choice_cols:

    for label_value in [0, 1]:

        subset = df_analysis[
            df_analysis[label_col] == label_value
        ]

        change_reports.append({
            "Môn lựa chọn": label_col,
            "Nhãn": label_value,
            "Số học sinh": len(subset),
            "Điểm TB lớp 6": subset["Lop_6"].mean(),
            "Điểm TB lớp 9": subset["Lop_9"].mean(),
            "Thay đổi 6-9": subset["Delta_6_9"].mean(),
            "Std thay đổi": subset["Delta_6_9"].std()
        })

change_df = lib.pd.DataFrame(change_reports).round(4)

print("SO SÁNH MỨC THAY ĐỔI ĐIỂM GIỮA CÁC NHÓM NHÃN")
display(change_df)

# ==========================================
# 4. BAR CHART - SO SÁNH MỨC THAY ĐỔI
# ==========================================

fig, ax = lib.plt.subplots(figsize=(15, 7))

lib.sns.barplot(
    data=change_df,
    x="Môn lựa chọn",
    y="Thay đổi 6-9",
    hue="Nhãn",
    hue_order=[0, 1],
    palette={0: "steelblue", 1: "darkorange"},
    ax=ax
)

ax.axhline(
    0,
    color="black",
    linewidth=1,
    linestyle="--"
)

ax.set_title(
    "So sánh mức thay đổi điểm lớp 6–9 theo nhóm nhãn",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Môn lựa chọn")
ax.set_ylabel("Thay đổi điểm trung bình (L9 - L6)")

ax.tick_params(axis="x", rotation=45)
ax.legend(
    title="Nhóm nhãn",
    labels=["Không chọn (0)", "Có chọn (1)"]
)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "learning_change_by_labels.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 5. LƯU KẾT QUẢ
# ==========================================

trajectory_df = lib.pd.concat(
    trajectory_reports,
    ignore_index=True
).round(4)

trajectory_df.to_csv(
    report_dir / "learning_trajectory_by_labels.csv",
    index=False,
    encoding="utf-8-sig"
)

change_df.to_csv(
    report_dir / "learning_change_by_labels.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```
