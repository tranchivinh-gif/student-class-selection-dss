```python
# Đọc dữ liệu Excel
file_path = "../data/data-clean/du_lieu_that.csv"
raw_df = lib.pd.read_csv(file_path)

choice_columns = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]
dfs = {
    f"Diem_L{grade}": raw_df[
        ["ma_hoc_sinh"]
        + [col for col in raw_df.columns if col.startswith(f"L{grade}_")]
        + choice_columns
    ].copy()
    for grade in [6, 7, 8, 9]
}

# Truy cập dữ liệu
df_l6 = dfs["Diem_L6"]
df_l7 = dfs["Diem_L7"]
df_l8 = dfs["Diem_L8"]
df_l9 = dfs["Diem_L9"]

# Hiển thị
display(df_l6.head())
display(df_l7.head())
display(df_l8.head())
display(df_l9.head())
```

```python
# ==========================================
# CORRELATION MATRIX - TƯƠNG QUAN GIỮA CÁC MÔN HỌC
# ==========================================

fig_dir = lib.Path("figures/correlation")
report_dir = lib.Path("reports/correlation")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

correlation_reports = {}

for sheet_name, df in dfs.items():

    # Chỉ lấy điểm cả năm của các môn học
    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col.endswith("_ca_nam")
        and col not in choice_cols
    ]

    # Tính ma trận tương quan Pearson
    corr_matrix = df[score_cols].corr(method="pearson")

    correlation_reports[sheet_name] = corr_matrix

    print(f"\n{'=' * 60}")
    print(f"SHEET: {sheet_name}")
    print(f"{'=' * 60}")

    display(corr_matrix.round(3))

    # ==========================================
    # HEATMAP - CORRELATION MATRIX
    # ==========================================

    fig, ax = lib.plt.subplots(figsize=(12, 9))

    mask = lib.np.triu(
        lib.np.ones_like(corr_matrix, dtype=bool),
        k=1
    )

    lib.sns.heatmap(
        corr_matrix,
        mask=mask,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        linewidths=0.5,
        cbar_kws={"label": "Pearson Correlation"},
        ax=ax
    )

    ax.set_title(
        f"Ma trận tương quan điểm cả năm - {sheet_name}",
        fontsize=14,
        fontweight="bold"
    )

    lib.plt.xticks(rotation=45, ha="right")
    lib.plt.yticks(rotation=0)

    lib.plt.tight_layout()

    # Lưu biểu đồ
    lib.plt.savefig(
        fig_dir / f"correlation_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

    # Lưu ma trận tương quan
    corr_matrix.to_csv(
        report_dir / f"correlation_{sheet_name}.csv",
        encoding="utf-8-sig"
    )

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# CORRELATION MATRIX
# PHÂN TÍCH TƯƠNG QUAN THEO HKI, HKII, CẢ NĂM
# ==========================================

fig_dir = lib.Path("figures/correlation")
report_dir = lib.Path("reports/correlation")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

periods = {
    "HKI": "_hki",
    "HKII": "_hkii",
    "CaNam": "_ca_nam"
}

correlation_reports = {}

for sheet_name, df in dfs.items():

    correlation_reports[sheet_name] = {}

    # Tạo 3 Heatmap cho mỗi lớp
    fig, axes = lib.plt.subplots(
        1, 3,
        figsize=(24, 7)
    )

    for ax, (period_name, suffix) in zip(
        axes, periods.items()
    ):

        # Lấy các cột điểm theo học kỳ
        score_cols = [
            col for col in df.select_dtypes(include="number").columns
            if col.endswith(suffix)
        ]

        # Tính tương quan Pearson
        corr_matrix = df[score_cols].corr(method="pearson")

        # Chuẩn hóa tên môn để dễ đọc
        corr_matrix.index = [
            col.removesuffix(suffix)
            for col in corr_matrix.index
        ]

        corr_matrix.columns = [
            col.removesuffix(suffix)
            for col in corr_matrix.columns
        ]

        # Lưu kết quả
        correlation_reports[sheet_name][period_name] = corr_matrix

        corr_matrix.to_csv(
            report_dir / f"correlation_{sheet_name}_{period_name}.csv",
            encoding="utf-8-sig"
        )

        # Chỉ hiển thị tam giác dưới
        mask = lib.np.triu(
            lib.np.ones_like(corr_matrix, dtype=bool),
            k=1
        )

        # Vẽ Heatmap
        lib.sns.heatmap(
            corr_matrix,
            mask=mask,
            annot=True,
            fmt=".2f",
            cmap="coolwarm",
            vmin=-1,
            vmax=1,
            center=0,
            square=True,
            linewidths=0.5,
            cbar_kws={"shrink": 0.7},
            ax=ax
        )

        ax.set_title(
            f"{sheet_name} - {period_name}",
            fontsize=13,
            fontweight="bold"
        )

        ax.tick_params(axis="x", rotation=45)
        ax.tick_params(axis="y", rotation=0)

    fig.suptitle(
        f"So sánh tương quan giữa các môn học - {sheet_name}",
        fontsize=16,
        fontweight="bold"
    )

    lib.plt.tight_layout()

    # Lưu 3 Heatmap trong cùng một ảnh
    lib.plt.savefig(
        fig_dir / f"correlation_3_periods_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# SCATTER PLOT - MỐI QUAN HỆ GIỮA ĐIỂM HAI MÔN
# ==========================================

fig_dir = lib.Path("figures/scatter")
fig_dir.mkdir(parents=True, exist_ok=True)

periods = {
    "HKI": "_hki",
    "HKII": "_hkii",
    "CaNam": "_ca_nam"
}

for sheet_name, df in dfs.items():

    for period_name, suffix in periods.items():

        # Lấy các cột điểm thuộc học kỳ tương ứng
        score_cols = [
            col for col in df.select_dtypes(include="number").columns
            if col.endswith(suffix)
        ]

        # Tính ma trận tương quan
        corr_matrix = df[score_cols].corr(method="pearson")

        # Lấy các cặp môn học không trùng nhau
        pairs = []

        for i in range(len(score_cols)):
            for j in range(i + 1, len(score_cols)):
                col_x = score_cols[i]
                col_y = score_cols[j]
                r = corr_matrix.loc[col_x, col_y]

                if lib.pd.notna(r):
                    pairs.append((col_x, col_y, r))

        # Chọn 3 cặp tương quan mạnh nhất và 3 cặp yếu nhất
        pairs_sorted = sorted(
            pairs,
            key=lambda x: abs(x[2]),
            reverse=True
        )

        selected_pairs = (
            pairs_sorted[:3] +
            pairs_sorted[-3:]
        )

        # Loại bỏ cặp bị lặp
        selected_pairs = list(dict.fromkeys(selected_pairs))

        if not selected_pairs:
            continue

        # Tạo lưới biểu đồ
        ncols = 3
        nrows = (len(selected_pairs) + ncols - 1) // ncols

        fig, axes = lib.plt.subplots(
            nrows, ncols,
            figsize=(18, nrows * 5)
        )

        axes = lib.np.array(axes).flatten()

        for i, (col_x, col_y, r) in enumerate(selected_pairs):

            ax = axes[i]

            # Vẽ Scatter Plot
            lib.sns.scatterplot(
                data=df,
                x=col_x,
                y=col_y,
                alpha=0.25,
                s=15,
                color="steelblue",
                ax=ax
            )

            # Đường xu hướng tuyến tính
            lib.sns.regplot(
                data=df,
                x=col_x,
                y=col_y,
                scatter=False,
                color="red",
                line_kws={"linewidth": 2},
                ci=None,
                ax=ax
            )

            ax.set_title(
                f"Pearson r = {r:.3f}",
                fontsize=12,
                fontweight="bold"
            )

            ax.set_xlabel(col_x.removesuffix(suffix))
            ax.set_ylabel(col_y.removesuffix(suffix))

            ax.set_xlim(0, 10)
            ax.set_ylim(0, 10)

        # Ẩn ô không sử dụng
        for j in range(len(selected_pairs), len(axes)):
            axes[j].axis("off")

        fig.suptitle(
            f"Scatter Plot - {sheet_name} - {period_name}",
            fontsize=16,
            fontweight="bold"
        )

        lib.plt.tight_layout()

        # Lưu biểu đồ
        lib.plt.savefig(
            fig_dir / f"scatter_{sheet_name}_{period_name}.png",
            dpi=300,
            bbox_inches="tight"
        )

        lib.plt.show()

print(f"Đã lưu các biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# BOXPLOT THEO NHÓM NHÃN (0/1)
# SO SÁNH PHÂN PHỐI ĐIỂM GIỮA HỌC SINH
# CHỌN VÀ KHÔNG CHỌN MÔN HỌC
# ==========================================

fig_dir = lib.Path("figures/boxplot_by_label")
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

periods = {
    "HKI": "_hki",
    "HKII": "_hkii",
    "CaNam": "_ca_nam"
}

# Lưu kết quả thống kê
boxplot_reports = {}

for sheet_name, df in dfs.items():

    for period_name, suffix in periods.items():

        # Lấy các cột điểm tương ứng
        score_cols = [
            col for col in df.select_dtypes(include="number").columns
            if col.endswith(suffix)
        ]

        # Tính điểm trung bình các môn trong kỳ
        # Đây là trung bình không trọng số của các cột điểm
        df_plot = df.copy()

        df_plot["diem_trung_binh"] = (
            df_plot[score_cols].mean(axis=1)
        )

        # Tạo lưới 3x3 cho 9 nhãn
        fig, axes = lib.plt.subplots(
            3, 3,
            figsize=(18, 15)
        )

        axes = axes.flatten()

        report_rows = []

        for i, label_col in enumerate(choice_cols):

            ax = axes[i]

            # Chỉ lấy học sinh có nhãn hợp lệ 0 hoặc 1
            plot_data = df_plot.loc[
                df_plot[label_col].isin([0, 1]),
                [label_col, "diem_trung_binh"]
            ].copy()

            # Vẽ Boxplot
            lib.sns.boxplot(
                data=plot_data,
                x=label_col,
                y="diem_trung_binh",
                order=[0, 1],
                color="skyblue",
                width=0.5,
                showfliers=True,
                ax=ax
            )

            ax.set_title(
                label_col,
                fontsize=12,
                fontweight="bold"
            )

            ax.set_xlabel("Lựa chọn môn học")
            ax.set_ylabel("Điểm trung bình")
            ax.set_ylim(0, 10)

            ax.set_xticks([0, 1])
            ax.set_xticklabels([
                "Không chọn (0)",
                "Có chọn (1)"
            ])

            # Thống kê theo từng nhóm nhãn
            for label_value in [0, 1]:

                values = plot_data.loc[
                    plot_data[label_col] == label_value,
                    "diem_trung_binh"
                ].dropna()

                report_rows.append({
                    "Sheet": sheet_name,
                    "Ky": period_name,
                    "Nhan": label_col,
                    "Gia_tri_nhan": label_value,
                    "So_hoc_sinh": len(values),
                    "Mean": values.mean(),
                    "Median": values.median(),
                    "Std": values.std(),
                    "Q1": values.quantile(0.25),
                    "Q3": values.quantile(0.75)
                })

        fig.suptitle(
            f"So sánh điểm theo nhóm nhãn - {sheet_name} - {period_name}",
            fontsize=16,
            fontweight="bold"
        )

        lib.plt.tight_layout()

        # Lưu biểu đồ
        lib.plt.savefig(
            fig_dir / f"boxplot_labels_{sheet_name}_{period_name}.png",
            dpi=300,
            bbox_inches="tight"
        )

        lib.plt.show()

        # Lưu kết quả thống kê
        report_df = lib.pd.DataFrame(report_rows).round(4)

        boxplot_reports[(sheet_name, period_name)] = report_df

        report_dir = lib.Path("reports/boxplot_by_label")
        report_dir.mkdir(parents=True, exist_ok=True)

        report_df.to_csv(
            report_dir / f"boxplot_stats_{sheet_name}_{period_name}.csv",
            index=False,
            encoding="utf-8-sig"
        )

        print(f"\n{sheet_name} - {period_name}")
        display(report_df)

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# BOXPLOT - ĐIỂM MÔN HỌC VÀ NHÃN LỰA CHỌN
# ==========================================

fig_dir = lib.Path("figures/subject_choice")
fig_dir.mkdir(parents=True, exist_ok=True)

# Ánh xạ nhãn lựa chọn với phần tên cột điểm tương ứng
subject_mapping = {
    "Vật lí": "khtn_ca_nam",
    "Hóa học": "khtn_ca_nam",
    "Sinh học": "khtn_ca_nam",
    "Địa lí": "lich_su_dia_li_ca_nam",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)": "gdcd_ca_nam",
    "Tin học": "tin_hoc_ca_nam",
    "Công nghệ": "cong_nghe_ca_nam"
}

for sheet_name, df in dfs.items():

    fig, axes = lib.plt.subplots(
        3, 3,
        figsize=(18, 15)
    )

    axes = axes.flatten()
    grade = sheet_name.rsplit("_L", 1)[-1]
    score_prefix = f"L{grade}_"

    for i, (label_col, score_col) in enumerate(
        subject_mapping.items()
    ):

        ax = axes[i]
        score_col = f"{score_prefix}{score_col}"

        missing_cols = [
            col for col in (label_col, score_col)
            if col not in df.columns
        ]
        if missing_cols:
            ax.axis("off")
            print(
                f"Bỏ qua {sheet_name} - {label_col}: "
                f"thiếu cột {missing_cols}"
            )
            continue

        plot_data = df[
            [label_col, score_col]
        ].dropna()

        plot_data = plot_data[
            plot_data[label_col].isin([0, 1])
        ]

        lib.sns.boxplot(
            data=plot_data,
            x=label_col,
            y=score_col,
            order=[0, 1],
            color="skyblue",
            width=0.5,
            ax=ax
        )

        ax.set_title(
            label_col,
            fontsize=12,
            fontweight="bold"
        )

        ax.set_xlabel("Lựa chọn môn học")
        ax.set_ylabel("Điểm môn học tương ứng")
        ax.set_ylim(0, 10)

        ax.set_xticks([0, 1])
        ax.set_xticklabels([
            "Không chọn (0)",
            "Có chọn (1)"
        ])

    # Ẩn 2 biểu đồ không sử dụng
    for j in range(len(subject_mapping), len(axes)):
        axes[j].axis("off")

    fig.suptitle(
        f"Mối quan hệ giữa điểm môn học và lựa chọn - {sheet_name}",
        fontsize=16,
        fontweight="bold"
    )

    lib.plt.tight_layout()

    lib.plt.savefig(
        fig_dir / f"subject_choice_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# POINT-BISERIAL CORRELATION
# MỐI LIÊN HỆ GIỮA ĐIỂM SỐ VÀ LỰA CHỌN MÔN
# ==========================================

fig_dir = lib.Path("figures/learning_choice")
report_dir = lib.Path("reports/learning_choice")

fig_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

periods = {
    "HKI": "_hki",
    "HKII": "_hkii",
    "CaNam": "_ca_nam"
}

correlation_results = {}

for sheet_name, df in dfs.items():

    for period_name, suffix in periods.items():

        # Lấy các cột điểm
        score_cols = [
            col for col in df.select_dtypes(include="number").columns
            if col.endswith(suffix)
        ]

        # Ma trận tương quan: điểm số × nhãn lựa chọn
        corr_matrix = lib.pd.DataFrame(
            index=score_cols,
            columns=choice_cols,
            dtype=float
        )

        pvalue_matrix = corr_matrix.copy()

        for score_col in score_cols:
            for choice_col in choice_cols:

                # Chỉ sử dụng các quan sát hợp lệ
                valid = df[
                    [score_col, choice_col]
                ].dropna()

                valid = valid[
                    valid[choice_col].isin([0, 1])
                ]

                if (
                    len(valid) >= 3
                    and valid[score_col].nunique() > 1
                    and valid[choice_col].nunique() == 2
                ):
                    r, p = lib.stats.pointbiserialr(
                        valid[choice_col],
                        valid[score_col]
                    )

                    corr_matrix.loc[score_col, choice_col] = r
                    pvalue_matrix.loc[score_col, choice_col] = p

        # Rút gọn tên cột điểm
        corr_matrix.index = [
            col.removesuffix(suffix)
            for col in corr_matrix.index
        ]

        pvalue_matrix.index = corr_matrix.index

        correlation_results[(sheet_name, period_name)] = corr_matrix

        # Lưu kết quả
        corr_matrix.to_csv(
            report_dir / f"correlation_{sheet_name}_{period_name}.csv",
            encoding="utf-8-sig"
        )

        pvalue_matrix.to_csv(
            report_dir / f"pvalue_{sheet_name}_{period_name}.csv",
            encoding="utf-8-sig"
        )

        # ==========================================
        # HEATMAP
        # ==========================================

        fig, ax = lib.plt.subplots(figsize=(16, 9))

        lib.sns.heatmap(
            corr_matrix,
            annot=True,
            fmt=".2f",
            cmap="coolwarm",
            center=0,
            vmin=-1,
            vmax=1,
            linewidths=0.5,
            cbar_kws={
                "label": "Point-Biserial Correlation"
            },
            ax=ax
        )

        ax.set_title(
            f"Tương quan giữa điểm số và lựa chọn môn - "
            f"{sheet_name} - {period_name}",
            fontsize=14,
            fontweight="bold"
        )

        ax.set_xlabel("Môn học được lựa chọn")
        ax.set_ylabel("Môn học THCS")

        lib.plt.xticks(rotation=45, ha="right")
        lib.plt.yticks(rotation=0)

        lib.plt.tight_layout()

        lib.plt.savefig(
            fig_dir / f"heatmap_{sheet_name}_{period_name}.png",
            dpi=300,
            bbox_inches="tight"
        )

        lib.plt.show()

        print(f"\n{'=' * 60}")
        print(f"{sheet_name} - {period_name}")
        print(f"{'=' * 60}")

        display(corr_matrix.round(3))

print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```
