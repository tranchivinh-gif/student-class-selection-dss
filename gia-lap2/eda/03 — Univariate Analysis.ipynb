```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import src.imports as lib
```

```python
# Đọc dữ liệu Excel
file_path = "../data/data-clean/du_lieu_that.csv"
raw_df = lib.pd.read_csv(file_path)

choice_columns = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]
dfs = {
    f"Diem_L{grade}": raw_df[
        ["ma_hoc_sinh"]
        + [col for col in raw_df.columns if col.startswith(f"L{grade}_")]
        + choice_columns
    ].copy()
    for grade in [6, 7, 8, 9]
}

# Truy cập dữ liệu
df_l6 = dfs["Diem_L6"]
df_l7 = dfs["Diem_L7"]
df_l8 = dfs["Diem_L8"]
df_l9 = dfs["Diem_L9"]

# Hiển thị
display(df_l6.head())
display(df_l7.head())
display(df_l8.head())
display(df_l9.head())
```

```python
# ==========================================
# HISTOGRAM - PHÂN PHỐI ĐIỂM THEO TỪNG LỚP
# ==========================================

fig_dir = lib.Path("figures/distribution")
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

for sheet_name, df in dfs.items():

    # Chỉ lấy các cột điểm
    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col not in choice_cols and col != "ma_hoc_sinh"
    ]

    # Tạo lưới biểu đồ
    ncols = 3
    nrows = (len(score_cols) + ncols - 1) // ncols

    fig, axes = lib.plt.subplots(
        nrows, ncols,
        figsize=(18, nrows * 3.5)
    )

    axes = lib.np.array(axes).flatten()

    for i, col in enumerate(score_cols):

        lib.sns.histplot(
            data=df,
            x=col,
            bins=20,
            kde=False,
            color="steelblue",
            edgecolor="white",
            ax=axes[i]
        )

        # Đường trung bình và trung vị
        mean_value = df[col].mean()
        median_value = df[col].median()

        axes[i].axvline(
            mean_value,
            color="red",
            linestyle="--",
            label=f"Mean: {mean_value:.2f}"
        )

        axes[i].axvline(
            median_value,
            color="green",
            linestyle=":",
            label=f"Median: {median_value:.2f}"
        )

        axes[i].set_title(col, fontsize=11)
        axes[i].set_xlabel("Điểm")
        axes[i].set_ylabel("Tần suất")
        axes[i].set_xlim(0, 10)
        axes[i].legend(fontsize=8)

    # Ẩn ô biểu đồ không sử dụng
    for j in range(len(score_cols), len(axes)):
        axes[j].axis("off")

    fig.suptitle(
        f"Phân phối điểm học sinh - {sheet_name}",
        fontsize=16,
        fontweight="bold"
    )

    lib.plt.tight_layout()

    # Lưu biểu đồ
    lib.plt.savefig(
        fig_dir / f"histogram_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

print(f"Đã lưu các biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# BOXPLOT - PHÁT HIỆN GIÁ TRỊ BẤT THƯỜNG
# ==========================================

fig_dir = lib.Path("figures/boxplot")
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

for sheet_name, df in dfs.items():

    # Chỉ lấy các cột điểm
    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col not in choice_cols and col != "ma_hoc_sinh"
    ]

    # Tạo lưới biểu đồ
    ncols = 3
    nrows = (len(score_cols) + ncols - 1) // ncols

    fig, axes = lib.plt.subplots(
        nrows, ncols,
        figsize=(18, nrows * 3.5)
    )

    axes = lib.np.array(axes).flatten()

    for i, col in enumerate(score_cols):
        lib.sns.boxplot(
            y=df[col],
            ax=axes[i],
            color="skyblue",
            width=0.4,
            fliersize=3
        )

        axes[i].set_title(col, fontsize=11)
        axes[i].set_ylabel("Điểm")
        axes[i].set_ylim(0, 10)

    # Ẩn ô thừa
    for j in range(len(score_cols), len(axes)):
        axes[j].axis("off")

    fig.suptitle(
        f"Boxplot phát hiện giá trị bất thường - {sheet_name}",
        fontsize=16,
        fontweight="bold"
    )

    lib.plt.tight_layout()

    # Lưu biểu đồ
    lib.plt.savefig(
        fig_dir / f"boxplot_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

print(f"Đã lưu các biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# KDE PLOT - HÌNH DẠNG PHÂN PHỐI ĐIỂM
# ==========================================

fig_dir = lib.Path("figures/kde")
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

for sheet_name, df in dfs.items():

    # Chỉ lấy các cột điểm
    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col not in choice_cols and col != "ma_hoc_sinh"
    ]

    # Tạo lưới biểu đồ
    ncols = 3
    nrows = (len(score_cols) + ncols - 1) // ncols

    fig, axes = lib.plt.subplots(
        nrows, ncols,
        figsize=(18, nrows * 3.5)
    )

    axes = lib.np.array(axes).flatten()

    for i, col in enumerate(score_cols):

        # Loại bỏ giá trị thiếu khi vẽ
        values = df[col].dropna()

        if len(values) >= 2 and values.nunique() > 1:
            lib.sns.kdeplot(
                x=values,
                fill=True,
                color="steelblue",
                alpha=0.4,
                linewidth=2,
                cut=0,
                clip=(0, 10),
                ax=axes[i]
            )

            # Đường trung bình và trung vị
            mean_value = values.mean()
            median_value = values.median()

            axes[i].axvline(
                mean_value,
                color="red",
                linestyle="--",
                label=f"Mean: {mean_value:.2f}"
            )

            axes[i].axvline(
                median_value,
                color="green",
                linestyle=":",
                label=f"Median: {median_value:.2f}"
            )

            axes[i].legend(fontsize=8)
        else:
            axes[i].text(
                0.5, 0.5,
                "Không đủ dữ liệu để vẽ KDE",
                ha="center",
                va="center",
                transform=axes[i].transAxes
            )

        axes[i].set_title(col, fontsize=11)
        axes[i].set_xlabel("Điểm")
        axes[i].set_ylabel("Mật độ")
        axes[i].set_xlim(0, 10)

    # Ẩn các ô không sử dụng
    for j in range(len(score_cols), len(axes)):
        axes[j].axis("off")

    fig.suptitle(
        f"KDE - Phân phối điểm học sinh - {sheet_name}",
        fontsize=16,
        fontweight="bold"
    )

    lib.plt.tight_layout()

    # Lưu biểu đồ
    lib.plt.savefig(
        fig_dir / f"kde_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()

print(f"Đã lưu các biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# DESCRIPTIVE STATISTICS
# Mean, Median, Std, Skewness
# ==========================================

report_dir = lib.Path("reports/descriptive_statistics")
report_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

descriptive_reports = {}

for sheet_name, df in dfs.items():

    # Chỉ lấy các cột điểm
    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col not in choice_cols and col != "ma_hoc_sinh"
    ]

    # Tính các chỉ số thống kê
    stats_df = lib.pd.DataFrame({
        "Count": df[score_cols].count(),
        "Mean": df[score_cols].mean(),
        "Median": df[score_cols].median(),
        "Std": df[score_cols].std(),
        "Skewness": df[score_cols].skew(),
        "Min": df[score_cols].min(),
        "Max": df[score_cols].max()
    }).round(4)

    stats_df.index.name = "Môn học / Học kỳ"

    # Lưu kết quả
    descriptive_reports[sheet_name] = stats_df

    stats_df.to_csv(
        report_dir / f"descriptive_{sheet_name}.csv",
        encoding="utf-8-sig"
    )

    # Hiển thị
    print(f"\n{'=' * 60}")
    print(f"SHEET: {sheet_name}")
    print(f"{'=' * 60}")

    display(stats_df)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# SO SÁNH PHÂN PHỐI ĐIỂM GIỮA CÁC LỚP 6–9
# ==========================================

fig_dir = lib.Path("figures/grade_comparison")
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = {
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
}

# Lấy HỢP tất cả các cột, không yêu cầu giống nhau giữa các lớp
all_cols = []
for df in dfs.values():
    for col in df.columns:
        if col not in all_cols:
            all_cols.append(col)

# Chuyển các cột có thể là điểm sang dạng số
for df in dfs.values():
    for col in df.columns:
        if col == "ma_hoc_sinh":
            continue

        converted = lib.pd.to_numeric(
            df[col]
            .astype("string")
            .str.strip()
            .str.replace(",", ".", regex=False),
            errors="coerce"
        )

        # Chỉ chuyển nếu cột có ít nhất một giá trị số
        if converted.notna().any():
            df[col] = converted

# Chọn các cột có dữ liệu số ở ít nhất một lớp
score_cols = []

for col in all_cols:
    if col in choice_cols or col == "ma_hoc_sinh":
        continue

    exists_as_numeric = any(
        col in df.columns
        and lib.pd.api.types.is_numeric_dtype(df[col])
        and df[col].notna().any()
        for df in dfs.values()
    )

    if exists_as_numeric:
        score_cols.append(col)

print("Các cột điểm được sử dụng:")
print(score_cols)

if len(score_cols) == 0:
    raise ValueError(
        "Không tìm thấy dữ liệu điểm dạng số. "
        "Hãy kiểm tra tên cột và dữ liệu trong dfs."
    )

# Gộp dữ liệu; lớp nào thiếu cột sẽ tự động nhận NaN
comparison_parts = []

for sheet_name, df in dfs.items():
    part = df.reindex(columns=score_cols).copy()
    part["Lop"] = sheet_name.replace("Diem_L", "Lớp ")
    comparison_parts.append(part)

comparison_df = lib.pd.concat(
    comparison_parts,
    ignore_index=True
)

grade_order = ["Lớp 6", "Lớp 7", "Lớp 8", "Lớp 9"]

# ==========================================
# 1. KDE
# ==========================================

ncols = 3
nrows = max(1, (len(score_cols) + ncols - 1) // ncols)

fig, axes = lib.plt.subplots(
    nrows,
    ncols,
    figsize=(18, nrows * 3.5),
    squeeze=False
)

axes = axes.ravel()

for i, col in enumerate(score_cols):
    has_plot = False

    for grade in grade_order:
        values = comparison_df.loc[
            comparison_df["Lop"] == grade,
            col
        ].dropna()

        if len(values) >= 2 and values.nunique() > 1:
            lib.sns.kdeplot(
                x=values,
                label=grade,
                linewidth=2,
                cut=0,
                clip=(0, 10),
                ax=axes[i]
            )
            has_plot = True

    axes[i].set_title(col, fontsize=11)
    axes[i].set_xlabel("Điểm")
    axes[i].set_ylabel("Mật độ")
    axes[i].set_xlim(0, 10)

    if has_plot:
        axes[i].legend(fontsize=8)

for j in range(len(score_cols), len(axes)):
    axes[j].axis("off")

fig.suptitle(
    "So sánh phân phối điểm giữa lớp 6, 7, 8 và 9",
    fontsize=16,
    fontweight="bold"
)

lib.plt.tight_layout()
lib.plt.savefig(
    fig_dir / "kde_grade_comparison.png",
    dpi=300,
    bbox_inches="tight"
)
lib.plt.show()

# ==========================================
# 2. BOXPLOT
# ==========================================

comparison_df["diem_trung_binh"] = comparison_df[score_cols].mean(axis=1)

fig, ax = lib.plt.subplots(figsize=(10, 6))

lib.sns.boxplot(
    data=comparison_df,
    x="Lop",
    y="diem_trung_binh",
    order=grade_order,
    color="skyblue",
    ax=ax
)

ax.set_title(
    "So sánh điểm trung bình giữa các lớp",
    fontsize=14,
    fontweight="bold"
)
ax.set_xlabel("Khối lớp")
ax.set_ylabel("Điểm trung bình")
ax.set_ylim(0, 10)

lib.plt.tight_layout()
lib.plt.savefig(
    fig_dir / "boxplot_grade_comparison.png",
    dpi=300,
    bbox_inches="tight"
)
lib.plt.show()

# ==========================================
# 3. BAR CHART
# ==========================================

summary_df = (
    comparison_df.groupby("Lop")["diem_trung_binh"]
    .agg(["mean", "median", "std"])
    .reindex(grade_order)
    .round(4)
)

display(summary_df)

plot_df = (
    summary_df[["mean", "median"]]
    .reset_index()
    .melt(
        id_vars="Lop",
        var_name="Thống kê",
        value_name="Điểm"
    )
    .dropna(subset=["Điểm"])
)

fig, ax = lib.plt.subplots(figsize=(10, 6))

lib.sns.barplot(
    data=plot_df,
    x="Lop",
    y="Điểm",
    hue="Thống kê",
    ax=ax
)

ax.set_title(
    "So sánh Mean và Median giữa các lớp",
    fontsize=14,
    fontweight="bold"
)
ax.set_xlabel("Khối lớp")
ax.set_ylabel("Điểm")
ax.set_ylim(0, 10)

lib.plt.tight_layout()
lib.plt.savefig(
    fig_dir / "mean_median_grade_comparison.png",
    dpi=300,
    bbox_inches="tight"
)
lib.plt.show()

print(f"Đã lưu 3 biểu đồ tại: {fig_dir.resolve()}")
```
