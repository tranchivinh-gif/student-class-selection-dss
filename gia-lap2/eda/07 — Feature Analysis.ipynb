```python
# Đọc dữ liệu CSV dạng rộng và tách theo từng lớp
file_path = "../data/data-clean/du_lieu_that.csv"
raw_df = lib.pd.read_csv(file_path)

choice_columns = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]
dfs = {
    f"Diem_L{grade}": raw_df[
        ["ma_hoc_sinh"]
        + [col for col in raw_df.columns if col.startswith(f"L{grade}_")]
        + choice_columns
    ].copy()
    for grade in [6, 7, 8, 9]
}

# Truy cập dữ liệu
df_l6 = dfs["Diem_L6"]
df_l7 = dfs["Diem_L7"]
df_l8 = dfs["Diem_L8"]
df_l9 = dfs["Diem_L9"]

# Hiển thị
display(df_l6.head())
display(df_l7.head())
display(df_l8.head())
display(df_l9.head())
```

```python
# ==========================================
# FEATURE ANALYSIS
# KIỂM TRA ĐẶC TRƯNG CÓ PHƯƠNG SAI THẤP
# ==========================================

report_dir = lib.Path("reports/feature_analysis")
fig_dir = lib.Path("figures/feature_analysis")

report_dir.mkdir(parents=True, exist_ok=True)
fig_dir.mkdir(parents=True, exist_ok=True)

# Chỉ kiểm tra 24 cột điểm, không bao gồm ID và 9 nhãn
variance_threshold = 0.01  # Ngưỡng tham khảo

variance_reports = []

for sheet_name, df in dfs.items():

    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col.endswith(("_hki", "_hkii", "_ca_nam"))
    ]

    # Tính phương sai từng đặc trưng
    variances = df[score_cols].var(ddof=0)

    report = lib.pd.DataFrame({
        "Sheet": sheet_name,
        "Đặc trưng": variances.index,
        "Phương sai": variances.values
    })

    report["Phương sai thấp"] = (
        report["Phương sai"] < variance_threshold
    )

    variance_reports.append(report)

# ==========================================
# 1. TỔNG HỢP KẾT QUẢ
# ==========================================

variance_df = lib.pd.concat(
    variance_reports,
    ignore_index=True
)

variance_df = variance_df.sort_values(
    ["Sheet", "Phương sai"]
).reset_index(drop=True)

print("=" * 60)
print("KIỂM TRA PHƯƠNG SAI CÁC ĐẶC TRƯNG")
print("=" * 60)

display(variance_df.round(4))

# ==========================================
# 2. LIỆT KÊ ĐẶC TRƯNG PHƯƠNG SAI THẤP
# ==========================================

low_variance_df = variance_df[
    variance_df["Phương sai thấp"]
]

print(f"Ngưỡng phương sai: {variance_threshold}")
print(f"Tổng số đặc trưng được kiểm tra: {len(variance_df)}")
print(f"Số đặc trưng có phương sai thấp: {len(low_variance_df)}")

if low_variance_df.empty:
    print("PASS: Không phát hiện đặc trưng có phương sai thấp.")
else:
    display(low_variance_df)

# ==========================================
# 3. HEATMAP PHƯƠNG SAI THEO LỚP
# ==========================================

heatmap_df = variance_df.pivot(
    index="Đặc trưng",
    columns="Sheet",
    values="Phương sai"
)

fig, ax = lib.plt.subplots(figsize=(10, 12))

lib.sns.heatmap(
    heatmap_df,
    annot=True,
    fmt=".3f",
    cmap="YlOrRd",
    linewidths=0.5,
    cbar_kws={"label": "Phương sai"},
    ax=ax
)

ax.set_title(
    "Phương sai đặc trưng điểm số theo khối lớp",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Khối lớp")
ax.set_ylabel("Đặc trưng")

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "feature_variance_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 4. LƯU BÁO CÁO
# ==========================================

variance_df.to_csv(
    report_dir / "feature_variance_report.csv",
    index=False,
    encoding="utf-8-sig"
)

low_variance_df.to_csv(
    report_dir / "low_variance_features.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# FEATURE ANALYSIS
# PHÁT HIỆN ĐẶC TRƯNG TRÙNG LẶP VÀ TƯƠNG QUAN CAO
# CHỈ SO SÁNH GIỮA CÁC MÔN HỌC KHÁC NHAU
# ==========================================

report_dir = lib.Path("reports/feature_analysis")
fig_dir = lib.Path("figures/feature_analysis")

report_dir.mkdir(parents=True, exist_ok=True)
fig_dir.mkdir(parents=True, exist_ok=True)

corr_threshold = 0.90

duplicate_reports = []
high_corr_reports = []

suffixes = ("_hki", "_hkii", "_ca_nam")


# ==========================================
# 1. HÀM XÁC ĐỊNH MÔN HỌC
# ==========================================

def get_subject(col):
    for suffix in suffixes:
        if col.endswith(suffix):
            return col.removesuffix(suffix)
    return col


# ==========================================
# 2. KIỂM TRA TỪNG SHEET
# ==========================================

for sheet_name, df in dfs.items():

    # Chỉ lấy các cột điểm
    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col.endswith(suffixes)
    ]

    X = df[score_cols].copy()

    # Tính ma trận tương quan Pearson
    corr_matrix = X.corr(method="pearson")

    for i in range(len(score_cols)):
        for j in range(i + 1, len(score_cols)):

            col1 = score_cols[i]
            col2 = score_cols[j]

            # Bỏ qua mọi cặp thuộc cùng một môn học
            if get_subject(col1) == get_subject(col2):
                continue

            # Kiểm tra trùng lặp hoàn toàn
            if X[col1].equals(X[col2]):
                duplicate_reports.append({
                    "Sheet": sheet_name,
                    "Đặc trưng 1": col1,
                    "Đặc trưng 2": col2
                })

            # Kiểm tra tương quan cao
            r = corr_matrix.loc[col1, col2]

            if lib.pd.notna(r) and abs(r) >= corr_threshold:
                high_corr_reports.append({
                    "Sheet": sheet_name,
                    "Đặc trưng 1": col1,
                    "Đặc trưng 2": col2,
                    "Pearson r": round(r, 4),
                    "|r|": round(abs(r), 4)
                })


# ==========================================
# 3. TỔNG HỢP KẾT QUẢ
# ==========================================

duplicate_df = lib.pd.DataFrame(
    duplicate_reports,
    columns=[
        "Sheet",
        "Đặc trưng 1",
        "Đặc trưng 2"
    ]
)

high_corr_df = lib.pd.DataFrame(
    high_corr_reports,
    columns=[
        "Sheet",
        "Đặc trưng 1",
        "Đặc trưng 2",
        "Pearson r",
        "|r|"
    ]
)

if not high_corr_df.empty:
    high_corr_df = high_corr_df.sort_values(
        "|r|",
        ascending=False
    ).reset_index(drop=True)


# ==========================================
# 4. HIỂN THỊ BÁO CÁO
# ==========================================

print("=" * 60)
print("KẾT QUẢ KIỂM TRA ĐẶC TRƯNG GIỮA CÁC MÔN")
print("=" * 60)

print(f"Ngưỡng tương quan: |r| >= {corr_threshold}")
print(f"Số cặp trùng lặp hoàn toàn: {len(duplicate_df)}")
print(f"Số cặp tương quan cao: {len(high_corr_df)}")

print("\nCÁC CẶP ĐẶC TRƯNG TRÙNG LẶP:")
display(duplicate_df)

print("\nCÁC CẶP ĐẶC TRƯNG TƯƠNG QUAN CAO:")
display(high_corr_df)


# ==========================================
# 5. TRỰC QUAN HÓA MA TRẬN TƯƠNG QUAN
# CHỈ GIỮ TƯƠNG QUAN GIỮA CÁC MÔN KHÁC NHAU
# ==========================================

for sheet_name, df in dfs.items():

    score_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col.endswith(suffixes)
    ]

    corr_matrix = df[score_cols].corr(method="pearson")

    # Ẩn các cặp cùng môn và đường chéo chính
    mask = lib.np.zeros(corr_matrix.shape, dtype=bool)

    for i, col1 in enumerate(score_cols):
        for j, col2 in enumerate(score_cols):
            if get_subject(col1) == get_subject(col2):
                mask[i, j] = True

    fig, ax = lib.plt.subplots(figsize=(16, 13))

    lib.sns.heatmap(
        corr_matrix,
        mask=mask,
        annot=False,
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        square=True,
        linewidths=0.3,
        cbar_kws={"label": "Pearson Correlation"},
        ax=ax
    )

    ax.set_title(
        f"Tương quan giữa các môn học khác nhau - {sheet_name}",
        fontsize=14,
        fontweight="bold"
    )

    lib.plt.xticks(rotation=90)
    lib.plt.yticks(rotation=0)

    lib.plt.tight_layout()

    lib.plt.savefig(
        fig_dir / f"cross_subject_correlation_{sheet_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    lib.plt.show()


# ==========================================
# 6. LƯU BÁO CÁO
# ==========================================

duplicate_df.to_csv(
    report_dir / "duplicate_features_cross_subject.csv",
    index=False,
    encoding="utf-8-sig"
)

high_corr_df.to_csv(
    report_dir / "high_correlation_cross_subject.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"\nĐã lưu báo cáo tại: {report_dir.resolve()}")
print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# FEATURE ANALYSIS
# PHÂN TÍCH MỐI LIÊN HỆ GIỮA ĐẶC TRƯNG VÀ BIẾN MỤC TIÊU
# POINT-BISERIAL CORRELATION
# ==========================================

report_dir = lib.Path("reports/feature_analysis")
fig_dir = lib.Path("figures/feature_analysis")

report_dir.mkdir(parents=True, exist_ok=True)
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

periods = {
    "HKI": "_hki",
    "HKII": "_hkii",
    "CaNam": "_ca_nam"
}

# ==========================================
# 1. GHÉP ĐẶC TRƯNG 4 LỚP VÀ 9 NHÃN
# ==========================================

X_parts = []

for grade in [6, 7, 8, 9]:

    df = dfs[f"Diem_L{grade}"]

    score_cols = [
        col for col in df.columns
        if col.endswith(("_hki", "_hkii", "_ca_nam"))
    ]

    temp = df[
        ["ma_hoc_sinh"] + score_cols
    ].copy()

    temp = temp.rename(
        columns={
            col: f"L{grade}_{col}"
            for col in score_cols
        }
    )

    X_parts.append(temp)

features_df = X_parts[0]

for temp in X_parts[1:]:
    features_df = features_df.merge(
        temp,
        on="ma_hoc_sinh",
        how="inner",
        validate="one_to_one"
    )

# Nhãn không đổi qua 4 lớp
labels_df = dfs["Diem_L9"][
    ["ma_hoc_sinh"] + choice_cols
].copy()

analysis_df = features_df.merge(
    labels_df,
    on="ma_hoc_sinh",
    how="inner",
    validate="one_to_one"
)

feature_cols = [
    col for col in analysis_df.columns
    if col.startswith(("L6_", "L7_", "L8_", "L9_"))
]

# ==========================================
# 2. TÍNH POINT-BISERIAL CORRELATION
# ==========================================

corr_matrix = lib.pd.DataFrame(
    index=feature_cols,
    columns=choice_cols,
    dtype=float
)

pvalue_matrix = corr_matrix.copy()

for feature in feature_cols:
    for target in choice_cols:

        valid = analysis_df[
            [feature, target]
        ].dropna()

        valid = valid[
            valid[target].isin([0, 1])
        ]

        if (
            len(valid) >= 3
            and valid[feature].nunique() > 1
            and valid[target].nunique() == 2
        ):
            r, p = lib.stats.pointbiserialr(
                valid[target],
                valid[feature]
            )

            corr_matrix.loc[feature, target] = r
            pvalue_matrix.loc[feature, target] = p

# ==========================================
# 3. TỔNG HỢP MỨC ĐỘ LIÊN HỆ
# ==========================================

correlation_report = (
    corr_matrix
    .stack()
    .reset_index()
)

correlation_report.columns = [
    "Đặc trưng",
    "Nhãn",
    "Pearson r"
]

correlation_report["|r|"] = (
    correlation_report["Pearson r"].abs()
)

correlation_report = correlation_report.sort_values(
    "|r|",
    ascending=False
).reset_index(drop=True)

print("=" * 60)
print("MỐI LIÊN HỆ GIỮA ĐẶC TRƯNG VÀ NHÃN")
print("=" * 60)

display(correlation_report.head(30))

# ==========================================
# 4. HEATMAP THEO TỪNG KHỐI LỚP
# ==========================================

for grade in [6, 7, 8, 9]:

    for period_name, suffix in periods.items():

        selected_cols = [
            col for col in feature_cols
            if col.startswith(f"L{grade}_")
            and col.endswith(suffix)
        ]

        plot_matrix = corr_matrix.loc[selected_cols].copy()

        plot_matrix.index = [
            col.replace(f"L{grade}_", "").removesuffix(suffix)
            for col in plot_matrix.index
        ]

        fig, ax = lib.plt.subplots(figsize=(15, 8))

        lib.sns.heatmap(
            plot_matrix,
            annot=True,
            fmt=".2f",
            cmap="coolwarm",
            center=0,
            vmin=-1,
            vmax=1,
            linewidths=0.5,
            cbar_kws={"label": "Point-Biserial r"},
            ax=ax
        )

        ax.set_title(
            f"Đặc trưng và nhãn - Lớp {grade} - {period_name}",
            fontsize=14,
            fontweight="bold"
        )

        ax.set_xlabel("Nhãn lựa chọn môn học")
        ax.set_ylabel("Đặc trưng điểm số")

        lib.plt.xticks(rotation=45, ha="right")
        lib.plt.yticks(rotation=0)

        lib.plt.tight_layout()

        lib.plt.savefig(
            fig_dir / f"feature_target_L{grade}_{period_name}.png",
            dpi=300,
            bbox_inches="tight"
        )

        lib.plt.show()

# ==========================================
# 5. TOP 15 ĐẶC TRƯNG LIÊN HỆ MẠNH NHẤT
# THEO TỪNG NHÃN
# ==========================================

top_features = []

for target in choice_cols:

    target_report = correlation_report[
        correlation_report["Nhãn"] == target
    ]

    top_features.append(
        target_report.head(15)
    )

top_features_df = lib.pd.concat(
    top_features,
    ignore_index=True
)

print("\nTOP 15 ĐẶC TRƯNG THEO TỪNG NHÃN")
display(top_features_df)

# ==========================================
# 6. LƯU BÁO CÁO
# ==========================================

corr_matrix.to_csv(
    report_dir / "feature_target_correlation_matrix.csv",
    encoding="utf-8-sig"
)

pvalue_matrix.to_csv(
    report_dir / "feature_target_pvalues.csv",
    encoding="utf-8-sig"
)

correlation_report.to_csv(
    report_dir / "feature_target_correlation.csv",
    index=False,
    encoding="utf-8-sig"
)

top_features_df.to_csv(
    report_dir / "top_features_by_target.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
print(f"Đã lưu biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# FEATURE ANALYSIS
# XÁC ĐỊNH ĐẶC TRƯNG KHÔNG ĐƯỢC SỬ DỤNG
# TẠI THỜI ĐIỂM DỰ ĐOÁN
# ==========================================

report_dir = lib.Path("reports/feature_analysis")
report_dir.mkdir(parents=True, exist_ok=True)

id_col = "ma_hoc_sinh"

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

# Thời điểm dự đoán: cuối lớp 6, 7, 8 hoặc 9
prediction_grades = [6, 7, 8, 9]

# Danh sách 96 đặc trưng điểm số
feature_cols = [
    col for col in analysis_df.columns
    if col.startswith(("L6_", "L7_", "L8_", "L9_"))
]

availability_reports = []

for prediction_grade in prediction_grades:

    for feature in feature_cols:

        feature_grade = int(feature.split("_")[0][1:])

        available = feature_grade <= prediction_grade

        availability_reports.append({
            "Thời điểm dự đoán": f"Cuối lớp {prediction_grade}",
            "Đặc trưng": feature,
            "Lớp chứa dữ liệu": feature_grade,
            "Được sử dụng": available,
            "Lý do": (
                "Đã có dữ liệu tại thời điểm dự đoán"
                if available
                else "Dữ liệu thuộc năm học tương lai"
            )
        })

availability_df = lib.pd.DataFrame(availability_reports)

# ==========================================
# 1. TỔNG HỢP THEO THỜI ĐIỂM DỰ ĐOÁN
# ==========================================

summary = (
    availability_df
    .groupby("Thời điểm dự đoán")["Được sử dụng"]
    .agg(
        So_dac_trung_duoc_dung="sum",
        Tong_dac_trung="count"
    )
    .reset_index()
)

summary["So_dac_trung_khong_duoc_dung"] = (
    summary["Tong_dac_trung"]
    - summary["So_dac_trung_duoc_dung"]
)

print("=" * 60)
print("KHẢ NĂNG SỬ DỤNG ĐẶC TRƯNG THEO THỜI ĐIỂM")
print("=" * 60)

display(summary)

# ==========================================
# 2. CÁC ĐẶC TRƯNG KHÔNG ĐƯỢC SỬ DỤNG
# ==========================================

unavailable_df = availability_df[
    ~availability_df["Được sử dụng"]
].copy()

display(unavailable_df)

# ==========================================
# 3. KIỂM TRA BIẾN ĐỊNH DANH VÀ BIẾN MỤC TIÊU
# ==========================================

excluded_cols = lib.pd.DataFrame([
    {
        "Biến": id_col,
        "Vai trò": "Định danh",
        "Được dùng làm X": False,
        "Lý do": "Chỉ dùng để liên kết hồ sơ học sinh"
    },
    *[
        {
            "Biến": col,
            "Vai trò": "Biến mục tiêu (y)",
            "Được dùng làm X": False,
            "Lý do": "Tránh rò rỉ nhãn mục tiêu"
        }
        for col in choice_cols
    ]
])

print("\nCÁC BIẾN LUÔN LOẠI KHỎI TẬP ĐẶC TRƯNG X")
display(excluded_cols)

# ==========================================
# 4. LƯU BÁO CÁO
# ==========================================

availability_df.to_csv(
    report_dir / "feature_availability.csv",
    index=False,
    encoding="utf-8-sig"
)

summary.to_csv(
    report_dir / "feature_availability_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

excluded_cols.to_csv(
    report_dir / "excluded_features.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# FEATURE ANALYSIS
# PHÁT HIỆN NGUY CƠ DATA LEAKAGE
# ==========================================

report_dir = lib.Path("reports/feature_analysis")
report_dir.mkdir(parents=True, exist_ok=True)

id_col = "ma_hoc_sinh"

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

# Thời điểm dự đoán
prediction_grades = [6, 7, 8, 9]

# Đặc trưng điểm số từ 4 lớp
feature_cols = [
    col for col in analysis_df.columns
    if col.startswith(("L6_", "L7_", "L8_", "L9_"))
]

leakage_reports = []

# ==========================================
# 1. KIỂM TRA RÒ RỈ THÔNG TIN TƯƠNG LAI
# ==========================================

for prediction_grade in prediction_grades:

    for feature in feature_cols:

        feature_grade = int(feature.split("_")[0][1:])

        if feature_grade > prediction_grade:
            leakage_reports.append({
                "Thời điểm": f"Cuối lớp {prediction_grade}",
                "Đặc trưng": feature,
                "Loại nguy cơ": "Future Data Leakage",
                "Mức độ": "Cao",
                "Nguyên nhân": "Sử dụng điểm của năm học tương lai"
            })

# ==========================================
# 2. KIỂM TRA RÒ RỈ NHÃN MỤC TIÊU
# ==========================================

for target in choice_cols:

    leakage_reports.append({
        "Thời điểm": "Tất cả",
        "Đặc trưng": target,
        "Loại nguy cơ": "Target Leakage",
        "Mức độ": "Nghiêm trọng",
        "Nguyên nhân": "Nhãn mục tiêu không được đưa vào X"
    })

# ==========================================
# 3. KIỂM TRA BIẾN ĐỊNH DANH
# ==========================================

leakage_reports.append({
    "Thời điểm": "Tất cả",
    "Đặc trưng": id_col,
    "Loại nguy cơ": "Identifier Leakage",
    "Mức độ": "Tiềm ẩn",
    "Nguyên nhân": "Mã học sinh chỉ dùng liên kết và chia dữ liệu"
})

# ==========================================
# 4. KIỂM TRA RÒ RỈ GIỮA TRAIN / TEST
# ==========================================

# Chỉ thực hiện nếu đã có train_df và test_df
if "train_df" in globals() and "test_df" in globals():

    train_ids = set(train_df[id_col].dropna())
    test_ids = set(test_df[id_col].dropna())

    overlapping_ids = train_ids.intersection(test_ids)

    if overlapping_ids:
        leakage_reports.append({
            "Thời điểm": "Train/Test",
            "Đặc trưng": id_col,
            "Loại nguy cơ": "Data Split Leakage",
            "Mức độ": "Nghiêm trọng",
            "Nguyên nhân": (
                f"{len(overlapping_ids)} học sinh xuất hiện ở cả train và test"
            )
        })

# ==========================================
# 5. TỔNG HỢP BÁO CÁO NGUY CƠ
# ==========================================

leakage_df = lib.pd.DataFrame(leakage_reports)

print("=" * 60)
print("BÁO CÁO NGUY CƠ DATA LEAKAGE")
print("=" * 60)

summary = (
    leakage_df
    .groupby(["Loại nguy cơ", "Mức độ"])
    .size()
    .reset_index(name="Số trường hợp")
)

display(summary)

print("\nCHI TIẾT CÁC NGUY CƠ:")
display(leakage_df)

# ==========================================
# 6. KIỂM TRA THỰC TẾ THEO THỜI ĐIỂM
# ==========================================

for prediction_grade in prediction_grades:

    valid_features = [
        col for col in feature_cols
        if int(col.split("_")[0][1:]) <= prediction_grade
    ]

    invalid_features = [
        col for col in feature_cols
        if int(col.split("_")[0][1:]) > prediction_grade
    ]

    print(f"\n{'=' * 60}")
    print(f"DỰ ĐOÁN CUỐI LỚP {prediction_grade}")
    print(f"{'=' * 60}")

    print(f"Đặc trưng hợp lệ: {len(valid_features)}")
    print(f"Đặc trưng tương lai cần loại khỏi X: {len(invalid_features)}")

# ==========================================
# 7. LƯU BÁO CÁO
# ==========================================

leakage_df.to_csv(
    report_dir / "data_leakage_risks.csv",
    index=False,
    encoding="utf-8-sig"
)

summary.to_csv(
    report_dir / "data_leakage_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

print(f"\nĐã lưu báo cáo tại: {report_dir.resolve()}")
```
