```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import src.imports as lib
```

```python
# Đọc dữ liệu CSV dạng rộng và tách theo từng lớp
file_path = "../data/data-clean/du_lieu_that.csv"
raw_df = lib.pd.read_csv(file_path)

choice_columns = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]
dfs = {
    f"Diem_L{grade}": raw_df[
        ["ma_hoc_sinh"]
        + [col for col in raw_df.columns if col.startswith(f"L{grade}_")]
        + choice_columns
    ].copy()
    for grade in [6, 7, 8, 9]
}

# Truy cập dữ liệu
df_l6 = dfs["Diem_L6"]
df_l7 = dfs["Diem_L7"]
df_l8 = dfs["Diem_L8"]
df_l9 = dfs["Diem_L9"]

# Hiển thị
display(df_l6.head())
display(df_l7.head())
display(df_l8.head())
display(df_l9.head())
```

```python
# ==========================================
# TARGET ANALYSIS
# ĐẾM SỐ LƯỢNG MẪU CỦA TỪNG LỚP NHÃN
# ==========================================

report_dir = lib.Path("reports/target_analysis")
report_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

# Nhãn không đổi qua các lớp nên chỉ lấy từ lớp 9
df_labels = dfs["Diem_L9"][choice_cols]

# Đếm số lượng mẫu của từng nhãn
label_counts = lib.pd.DataFrame({
    "Không chọn (0)": (df_labels == 0).sum(),
    "Có chọn (1)": (df_labels == 1).sum()
})

label_counts["Tổng mẫu"] = label_counts.sum(axis=1)

label_counts.index.name = "Môn học"

# Sắp xếp theo số lượng mẫu chọn môn
label_counts = label_counts.sort_values(
    "Có chọn (1)",
    ascending=False
)

print("=" * 60)
print("SỐ LƯỢNG MẪU CỦA TỪNG LỚP NHÃN")
print("=" * 60)

display(label_counts)

# Lưu kết quả
label_counts.to_csv(
    report_dir / "label_counts.csv",
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# TARGET ANALYSIS
# TÍNH TỶ LỆ PHẦN TRĂM CÁC LỚP NHÃN
# ==========================================

# Sử dụng label_counts đã tạo ở bước trước
label_percent = label_counts[
    ["Không chọn (0)", "Có chọn (1)"]
].div(
    label_counts["Tổng mẫu"], axis=0
).mul(100).round(2)

label_percent.columns = [
    "Tỷ lệ không chọn (%)",
    "Tỷ lệ có chọn (%)"
]

label_percent["Tổng tỷ lệ (%)"] = (
    label_percent.sum(axis=1)
)

print("=" * 60)
print("TỶ LỆ PHẦN TRĂM CÁC LỚP NHÃN")
print("=" * 60)

display(label_percent)

# Lưu kết quả
label_percent.to_csv(
    report_dir / "label_percentages.csv",
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# TARGET ANALYSIS
# COUNTPLOT - PHÂN BỐ NHÃN LỰA CHỌN MÔN HỌC
# ==========================================

fig_dir = lib.Path("figures/target_analysis")
fig_dir.mkdir(parents=True, exist_ok=True)

choice_cols = [
    "Vật lí", "Hóa học", "Sinh học", "Địa lí",
    "Giáo dục kinh tế và pháp luật (GDKT&PL)",
    "Tin học", "Công nghệ", "Âm nhạc", "Mĩ thuật"
]

# Nhãn không đổi qua các lớp nên chỉ sử dụng lớp 9
df_labels = dfs["Diem_L9"][choice_cols]

# Chuyển dữ liệu từ dạng wide sang long
plot_df = df_labels.melt(
    var_name="Môn học",
    value_name="Lựa chọn"
)

# Chỉ giữ nhãn hợp lệ
plot_df = plot_df[
    plot_df["Lựa chọn"].isin([0, 1])
].copy()

# ==========================================
# 1. COUNTPLOT - SO SÁNH NHÃN 0 VÀ 1
# ==========================================

fig, ax = lib.plt.subplots(figsize=(16, 8))

lib.sns.countplot(
    data=plot_df,
    x="Môn học",
    hue="Lựa chọn",
    order=choice_cols,
    hue_order=[0, 1],
    palette={0: "steelblue", 1: "darkorange"},
    ax=ax
)

# Hiển thị số lượng trên từng cột
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=3,
        fontsize=9
    )

ax.set_title(
    "Phân bố số lượng học sinh theo từng nhãn lựa chọn môn học",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel("Môn học")
ax.set_ylabel("Số học sinh")

ax.tick_params(axis="x", rotation=45)

ax.legend(
    title="Nhãn",
    labels=["Không chọn (0)", "Có chọn (1)"]
)

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "countplot_label_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

# ==========================================
# 2. COUNTPLOT - CHỈ NHÓM CÓ CHỌN (1)
# ==========================================

selected_df = plot_df[
    plot_df["Lựa chọn"] == 1
]

fig, ax = lib.plt.subplots(figsize=(14, 7))

lib.sns.countplot(
    data=selected_df,
    y="Môn học",
    order=selected_df["Môn học"].value_counts().index,
    color="steelblue",
    ax=ax
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=5
    )

ax.set_title(
    "Số lượng học sinh lựa chọn từng môn học",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel("Số học sinh")
ax.set_ylabel("Môn học")

lib.plt.tight_layout()

lib.plt.savefig(
    fig_dir / "countplot_selected_subjects.png",
    dpi=300,
    bbox_inches="tight"
)

lib.plt.show()

print(f"Đã lưu 2 biểu đồ tại: {fig_dir.resolve()}")
```

```python
# ==========================================
# TARGET ANALYSIS
# XÁC ĐỊNH MỨC ĐỘ MẤT CÂN BẰNG LỚP
# ==========================================

report_dir = lib.Path("reports/target_analysis")
report_dir.mkdir(parents=True, exist_ok=True)

# Sử dụng label_counts đã tạo ở bước trước
imbalance_df = label_counts[
    ["Không chọn (0)", "Có chọn (1)"]
].copy()

# Tính tỷ lệ phần trăm
imbalance_df["Tỷ lệ nhãn 0 (%)"] = (
    imbalance_df["Không chọn (0)"]
    / imbalance_df.sum(axis=1) * 100
).round(2)

imbalance_df["Tỷ lệ nhãn 1 (%)"] = (
    100 - imbalance_df["Tỷ lệ nhãn 0 (%)"]
).round(2)

# Imbalance Ratio = số mẫu lớp đa số / số mẫu lớp thiểu số
majority = imbalance_df[
    ["Không chọn (0)", "Có chọn (1)"]
].max(axis=1)

minority = imbalance_df[
    ["Không chọn (0)", "Có chọn (1)"]
].min(axis=1)

imbalance_df["Imbalance Ratio"] = (
    majority / minority.replace(0, float("nan"))
).round(2)

# Phân loại mức độ mất cân bằng theo ngưỡng tham khảo
def classify_imbalance(ratio):
    if lib.pd.isna(ratio):
        return "Chỉ có một lớp"
    elif ratio < 1.5:
        return "Cân bằng"
    elif ratio < 3:
        return "Mất cân bằng nhẹ"
    elif ratio < 10:
        return "Mất cân bằng trung bình"
    else:
        return "Mất cân bằng nghiêm trọng"

imbalance_df["Mức độ"] = (
    imbalance_df["Imbalance Ratio"]
    .apply(classify_imbalance)
)

# Sắp xếp theo mức độ mất cân bằng
imbalance_df = imbalance_df.sort_values(
    "Imbalance Ratio",
    ascending=False
)

print("=" * 60)
print("ĐÁNH GIÁ MỨC ĐỘ MẤT CÂN BẰNG CÁC NHÃN")
print("=" * 60)

display(imbalance_df)

# Lưu kết quả
imbalance_df.to_csv(
    report_dir / "label_imbalance_report.csv",
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```

```python
# ==========================================
# TARGET ANALYSIS
# KIỂM TRA CÁC LỚP NHÃN CÓ QUÁ ÍT MẪU
# ==========================================

report_dir = lib.Path("reports/target_analysis")
report_dir.mkdir(parents=True, exist_ok=True)

# Sử dụng label_counts đã tạo ở bước trước
min_samples = 100  # Ngưỡng tham khảo

rare_labels = label_counts[
    ["Không chọn (0)", "Có chọn (1)"]
].copy()

# Xác định số mẫu của lớp thiểu số
rare_labels["Số mẫu thiểu số"] = rare_labels[
    ["Không chọn (0)", "Có chọn (1)"]
].min(axis=1)

rare_labels["Tỷ lệ thiểu số (%)"] = (
    rare_labels["Số mẫu thiểu số"]
    / rare_labels[["Không chọn (0)", "Có chọn (1)"]].sum(axis=1)
    * 100
).round(2)

# Đánh dấu các nhãn có quá ít mẫu
rare_labels["Ít mẫu"] = (
    rare_labels["Số mẫu thiểu số"] < min_samples
)

rare_labels = rare_labels.sort_values(
    "Số mẫu thiểu số",
    ascending=True
)

print("=" * 60)
print("KIỂM TRA CÁC NHÃN CÓ QUÁ ÍT MẪU")
print("=" * 60)

display(rare_labels)

# ==========================================
# DANH SÁCH NHÃN CẦN CHÚ Ý
# ==========================================

rare_report = rare_labels[
    rare_labels["Ít mẫu"]
]

print(f"\nNgưỡng kiểm tra: {min_samples} mẫu")
print(f"Số nhãn có lớp thiểu số dưới ngưỡng: {len(rare_report)}")

if not rare_report.empty:
    print("\nCác nhãn cần chú ý:")
    display(rare_report)
else:
    print("PASS: Không có nhãn nào có lớp thiểu số dưới ngưỡng.")

# ==========================================
# LƯU BÁO CÁO
# ==========================================

rare_labels.to_csv(
    report_dir / "rare_label_analysis.csv",
    encoding="utf-8-sig"
)

print(f"Đã lưu báo cáo tại: {report_dir.resolve()}")
```
